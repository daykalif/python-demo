```python
# 数据选择

df4 = pd.read_csv('data/movie_list.csv', usecols=['电影名', '电影时长', '导演'], index_col='电影时长')

# 1.选择列
# 1.1 单列
df4['电影名']  # 显示单列数据

# 1.2 多列
df4[['电影名', '导演']]  # 显示多列数据

# 2.选择行 - iloc, loc
# 2.1 iloc ---> 基于行号选择行(不包含结束位置)，语法：df.iloc[start:stop:step]
df4.iloc[0:5:1]  # 显示前5行数据
df4.iloc[0:5]  # 显示前5行数据

# 2.2 loc ---> 基于行索引选择行(包含结束位置)，语法：df.loc[start:stop:step]
df4.loc[85:96:2]
```
