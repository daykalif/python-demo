```python
import pandas as pd

pd.set_option('display.min_rows', 30)  # 设置显示行数（默认10行）

# 1.读取数据
df = pd.read_csv('data/order.csv')

# 2.数据清洗
# 2.1 处理缺失值
df.isnull()  # 查看缺失值

# 2.1.1 
# 删除缺失值所在行
df.dropna()
# 删除缺失值所在列
df = df.dropna(axis=1)

# 2.1.2 
# 填充缺失值
df.fillna('-')

# 填充缺失值,填充上一行数据
df.ffill()
# 填充缺失值,填充下一行数据
df.bfill()

# 2.2 处理重复值
# 2.2.1 查看重复值
df.duplicated()  # 所有的列的数据都重复，才会判断为重复
df.duplicated(subset=['订单号'])  # 只要订单号重复，就会判断为重复

# 2.2.2 删除重复值
df.drop_duplicates()
df.drop_duplicates(subset=['订单号'])  # 删除重复值（只要订单号重复，就会判断为重复）
df.drop_duplicates(subset=['订单号'], keep='last')  # 删除重复值，只保留最后一个; 如果keep='first', 则保留第一个; 如果keep=False, 则删除所有重复值

# 2.3 处理异常值
# 2.3.1 查看异常值
df[df['单价'] < 0]

# 2.3.2 删除异常值
df.drop(df[df['单价'] < 0].index)

# 2.3.3 修复异常值
df['单价'] = df['单价'].abs()  # 绝对值

# 2.4 数据格式处理
df['订单日期'] = df['订单日期'].str.replace('/', '-')
df
```
