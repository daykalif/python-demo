```python
import pandas as pd

# 1.读取数据
df = pd.read_csv('data/order.csv', nrows=10)  # 读取前10行数据

# 2.排序
# 2.1 根据 销售数量 倒序排序
df.sort_values(by='销售数量', ascending=False)

# 2.2 根据 单价 升序排序
df.sort_values(by='单价', ascending=True)
df.sort_values(by='单价')  # ascending默认为True，表示升序排序

# 2.3 根据 单价 升序排序，价格一样，再根据 销售数量 倒序排序
df.sort_values(by=['单价', '销售数量'], ascending=[True, False])

df.sort_values(by=['单价', '销售数量'], ascending=True)  # 如果两列都要求升序排序，可以简写为True
```
