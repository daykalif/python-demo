```python
import pandas as pd

# 构造DataFrame -- 创建数据集（学员成绩信息）
df = pd.DataFrame([
    {"姓名": "小王", "语文": 90, "数学": 80, "英语": 70},
    {"姓名": "小李", "语文": 80, "数学": 70, "英语": 60},
    {"姓名": "小张", "语文": 70, "数学": 60, "英语": 50},
])

print(df)  # 运行快捷键：shift + Enter, 默认只会运行输出最后一行的内容，如果需要多行输出，可以使用print函数
df["语文"]
df["语文"].max(), df["语文"].min(), df["语文"].mean()
f"语文最高分: {df["语文"].max()},最低分: {df["语文"].min()}, 平均分: {df["语文"].mean()}"
```
