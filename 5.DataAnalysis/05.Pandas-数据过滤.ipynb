```python
"""
Pandas 数据查看选择与数据过滤操作

1. 数据的查看与选择
查看：head ()、tail ()、describe ()、info ()
选择列：df [' 列名 '] /df [[' 列名 1', ' 列名 2']]
选择行：df.iloc [start:stop:step] /df.loc [start:stop:step]

2. 数据的过滤操作
语法：df [条件表达式]
示例：df [df [' 单价 '] >= 100] /df [df [' 类别 '].isin (['...','...'])]
"""
import pandas as pd

df = pd.read_csv('data/sales.csv')

# 配置项，展示所有的数据
pd.set_option('display.max_rows', None)

# 数据过滤 ---> df[filter]
# 1. 销售数量 >= 10 的订单数据
df[df['销售数量'] >= 10]

# 2. 产品类别为 食品 或 图书 的订单数据
df[df['产品类别'].isin(['食品', '图书'])]  # 或：df[(df['产品类别'] == '食品') | (df['产品类别'] == '图书')]

# 3. 单价在 100-200 之间的订单数据（between：包含最小值，也包含最大值）
df[df['单价'].between(100, 200)]  # 或：df[(df['单价'] >= 100) & (df['单价'] <= 200)]
df[df['单价'].between(100, 200, inclusive='left')]  # inclusive: 'left' 表示包含左边界，'right' 表示包含右边界

# 4. 销售数量 >= 8，并且 单价 >= 100 的订单数据 ---> 多条件（并且：& ，或者：| ）
df[(df['销售数量'] >= 8) & (df['单价'] >= 100)]

# 5. 产品类别为 服装 / 食品，支付方式为 支付宝 / 微信支付 的订单数据
df[(df['产品类别'].isin(['服装', '食品'])) & (df['支付方式'].isin(['支付宝', '微信支付']))]
```
