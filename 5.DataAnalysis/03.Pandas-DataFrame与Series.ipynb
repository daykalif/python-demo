```python
import pandas as pd

# 方式1：列表嵌套字典创建DataFrame（一行数据对应一个字典）
df1 = pd.DataFrame([
    {'姓名': '王林', '语文': 80, '英语': 90, '数学': 88},
    {'姓名': '李慕婉', '语文': 92, '英语': 81, '数学': 93},
    {'姓名': '十三', '语文': 87, '英语': 83, '数学': 78}
])

# 方式2：字典嵌套列表创建DataFrame（一列数据对应一个列表）
df2 = pd.DataFrame({
    '姓名': ['王林', '李慕婉', '十三'],
    '语文': [80, 92, 87],
    '英语': [90, 81, 83],
    '数学': [88, 93, 78]
})

# df3：二维列表构造，通过columns指定列名，行索引默认0,1,2
df3 = pd.DataFrame([
    ['王林', 80, 90, 88],
    ['李慕婉', 92, 81, 93],
    ['十三', 87, 83, 78]
], columns=['姓名', '语文', '英语', '数学'])

# df4：二维列表构造，同时指定列名columns与自定义行索引index
df4 = pd.DataFrame([
    ['王林', 80, 90, 88],
    ['李慕婉', 92, 81, 93],
    ['十三', 87, 83, 78]
], columns=['姓名', '语文', '英语', '数学'], index=['a', 'b', 'c'])

print("===== df1（列表套字典） =====")
print(df1)
print("\n===== df2（字典套列表） =====")
print(df2)
print("===== df3（默认行索引） =====")
print(df3)
print("\n===== df4（自定义行索引abc） =====")
print(df4)

# DataFrame常见属性 - index,columns,values,size,dtypes,shape
df4.index.tolist()  # index: 索引
df4.columns.tolist()  # columns: 列名
df4.values.tolist()  # values: 值
df4.size  # size: 大小
df4.dtypes  # dtypes: 数据类型
df4.shape  # shape: 维度（行，列）

```

```python
"""
小结：

1.xx.index：获取行索引
Series：返回序列的行索引对象，比如RangeIndex(start=0, stop=5, step=1)这类默认自增索引，或是自定义标签索引。
DataFrame：返回表格的行索引，用来标识每一行数据。

2.xx.values：获取底层数据数组
Series：返回存放所有元素的一维 numpy 数组。
DataFrame：返回二维 numpy 数组，按行排布表格里的所有数值。

3.xx.dtype：获取数据类型
Series：返回整个序列元素的数据类型（如int64、object字符串型）。
DataFrame不能直接用 dtype，需要用df.dtypes（复数），会返回每一列各自的数据类型。

4.xx.size：获取数据总个数
Series：直接返回一维序列的元素总数量。
DataFrame：返回行数 × 列数的总单元格数量。

5.xx.shape：获取数据维度，格式为(行数, 列数)
Series：返回(元素总数,) 这样的一元元组，代表一维长度。
DataFrame：返回(行数量, 列数量)的二元元组，直观看出表格行列规模。

6.xx.columns：DataFrame 独有属性，获取列名
Series没有 columns 属性（一维结构不存在列的概念）；
DataFrame 返回所有列的索引对象，能拿到表格每一列的名称，也支持修改列名。
"""
```
